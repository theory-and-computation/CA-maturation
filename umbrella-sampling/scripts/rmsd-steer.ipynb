```python
import numpy as np
import matplotlib.pyplot as plt

dat_2d_0822 = np.loadtxt("output_0822_s.txt",delimiter=',')
dat_2d_0823 = np.loadtxt("rmsd_0823.txt",delimiter=',')
dat_2d_0912 = np.loadtxt("rmsd_0912.txt",delimiter=',')
dat_2d_0920 = np.loadtxt("rmsd_0920.txt",delimiter=',')

dat_2d =dat_2d_0822[np.argsort(dat_2d_0822[:,0])]
dat_2d = np.hstack((dat_2d,np.zeros((dat_2d.shape[0],3))))

for i in range(dat_2d.shape[0]):
    dat_2d[i,3] = dat_2d_0823[np.where(np.all(dat_2d_0823[:,:2]==dat_2d[i,:2],axis=1))[0][0],2]
    dat_2d[i,4] = dat_2d_0912[np.where(np.all(dat_2d_0912[:,:2]==dat_2d[i,:2],axis=1))[0][0],2]
    dat_2d[i,5] = dat_2d_0920[np.where(np.all(dat_2d_0920[:,:2]==dat_2d[i,:2],axis=1))[0][0],2]
 
for j in range(2,6):
    plt.figure()
    plt.scatter(dat_2d[:,0],dat_2d[:,1],c=dat_2d[:,j],cmap="RdBu")
    plt.colorbar()
```
