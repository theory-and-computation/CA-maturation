```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
```

```python

dat = np.loadtxt("rmsd_1111.txt",delimiter=",")

dat_morph = np.loadtxt("morph_data.txt",delimiter=",");
norm = mpl.colors.Normalize(vmin=np.min(dat[:,2]),vmax=np.max(dat[:,2]))

plt.scatter(dat[:,0],dat[:,1],c=dat[:,2],norm=norm,cmap="RdBu")
plt.xlabel("dihedral")
plt.ylabel("distance")
plt.colorbar()

plt.plot(dat_morph[2:,2],dat_morph[2:,1],c='k')
plt.scatter(dat_morph[:,2],dat_morph[:,1],c=dat_morph[:,3],marker='o',zorder=4,norm=norm,cmap="RdBu")
```
