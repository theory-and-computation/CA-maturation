```python
import numpy as np
import matplotlib.pyplot as plt

dat = np.loadtxt("CA-hex-pent/2-equil/energy_2_3.csv",delimiter=',')
half = int(len(dat)/2)
tm = dat[:,0] / (500000) # time in ns
tm[half:] += 5
plt.figure(figsize=(12,4))
plt.subplot(131)
plt.plot(tm[10:],dat[10:,10]) # total energy
plt.title("Total energy")
plt.xlabel("Time (ns)")
plt.subplot(132)
plt.plot(tm[10:],dat[10:,11]) # temp
plt.title("Temperature")
plt.xlabel("Time (ns)")
plt.subplot(133)
plt.plot(tm[10:],dat[10:,15]) # pressure
plt.title("Pressure")
plt.xlabel("Time (ns)")
```
